## ⚙️ Hapi 1: Setup-i dhe Libraritë
Në këtë pjesë do të importojmë të gjitha libraritë e nevojshme dhe do të konfigurojmë shtigjet e dosjeve në mënyrë që të punojnë si nga ky notebook ashtu edhe nga pjesa tjetër e projektit.

```python
import os
import pickle
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from time import time

# ML metrics & models
from sklearn.model_selection import train_test_split, GridSearchCV, cross_val_score, StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neural_network import MLPClassifier
from sklearn.cluster import KMeans
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, silhouette_score, adjusted_rand_score
from sklearn.decomposition import PCA

warnings.filterwarnings('ignore')

# Konfigurimi i shtigjeve robuste (përshtatet automatikisht)
if Path("ml").exists():
    BASE_DIR = Path(".")
else:
    BASE_DIR = Path("..")

RAW_DIR = BASE_DIR / "datasets"
PROCESSED_DIR = RAW_DIR / "processed"
MODELS_DIR = BASE_DIR / "models"
VIS_DIR = BASE_DIR / "visualizations"

# Krijimi i dosjeve nëse nuk ekzistojnë
RAW_DIR.mkdir(exist_ok=True)
PROCESSED_DIR.mkdir(exist_ok=True)
MODELS_DIR.mkdir(exist_ok=True)
VIS_DIR.mkdir(exist_ok=True)

# Rregullimi i stilit grafik të vizualizimeve
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("husl")
RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

print(f"[OK] Ambient u konfigurua me sukses!")
print(f"  BASE_DIR      : {BASE_DIR.resolve()}")
print(f"  DATASETS_DIR  : {RAW_DIR.resolve()}")
print(f"  MODELS_DIR    : {MODELS_DIR.resolve()}")
print(f"  VIS_DIR       : {VIS_DIR.resolve()}")
```

## 📊 Hapi 2: Gjenerimi i Dataset-it (Step 0)
Në mungesë të lidhjes direkte me Kaggle, ne do të simulojmë dataset-in e famshëm **Pima Indians Diabetes Database** me saktësi statistikore. Gjithashtu do të fusim vlera `0` si kode për vlerat e munguara (p.sh. Glukozë apo Insulinë zero) për të imituar saktësisht problemet e të dhënave klinike reale.

```python
def generate_diabetes_dataset(n_samples: int = 768) -> pd.DataFrame:
    n_negative = int(n_samples * 0.651)  # ~500 jo-diabetik
    n_positive = n_samples - n_negative   # ~268 diabetik
    
    # --- Pacientët pa diabet (Outcome=0) ---
    negative = pd.DataFrame({
        'Pregnancies': np.random.poisson(3.3, n_negative).clip(0, 17),
        'Glucose': np.random.normal(110, 26, n_negative).clip(44, 199).round().astype(int),
        'BloodPressure': np.random.normal(68, 18, n_negative).clip(0, 122).round().astype(int),
        'SkinThickness': np.random.normal(19.7, 14.9, n_negative).clip(0, 99).round().astype(int),
        'Insulin': np.random.gamma(2, 30, n_negative).clip(0, 846).round().astype(int),
        'BMI': np.random.normal(30.3, 7.7, n_negative).clip(0, 67.1).round(1),
        'DiabetesPedigreeFunction': np.random.gamma(2, 0.2, n_negative).clip(0.078, 2.42).round(3),
        'Age': np.random.gamma(2, 8, n_negative).clip(21, 81).round().astype(int),
        'Outcome': 0
    })
    
    # --- Pacientët me diabet (Outcome=1) ---
    positive = pd.DataFrame({
        'Pregnancies': np.random.poisson(4.9, n_positive).clip(0, 17),
        'Glucose': np.random.normal(141, 32, n_positive).clip(44, 199).round().astype(int),
        'BloodPressure': np.random.normal(70, 21, n_positive).clip(0, 122).round().astype(int),
        'SkinThickness': np.random.normal(22.2, 17.7, n_positive).clip(0, 99).round().astype(int),
        'Insulin': np.random.gamma(2.5, 40, n_positive).clip(0, 846).round().astype(int),
        'BMI': np.random.normal(35.1, 7.3, n_positive).clip(0, 67.1).round(1),
        'DiabetesPedigreeFunction': np.random.gamma(2.5, 0.22, n_positive).clip(0.078, 2.42).round(3),
        'Age': np.random.gamma(2.5, 10, n_positive).clip(21, 81).round().astype(int),
        'Outcome': 1
    })
    
    df = pd.concat([negative, positive], ignore_index=True)
    df = df.sample(frac=1, random_state=RANDOM_STATE).reset_index(drop=True)
    
    # Vendosja e vlerave 0 që tregojnë vlera të munguara (missing values)
    zero_cols = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
    zero_rates = [0.007, 0.046, 0.296, 0.487, 0.014]
    for col, rate in zip(zero_cols, zero_rates):
        n_zeros = int(len(df) * rate)
        idx = np.random.choice(df.index, n_zeros, replace=False)
        df.loc[idx, col] = 0
        
    return df

print("Po gjenerohet dataset-i i Diabetit...")
df_raw = generate_diabetes_dataset(768)
diabetes_csv_path = RAW_DIR / "diabetes.csv"
df_raw.to_csv(diabetes_csv_path, index=False)
print(f"  [OK] Dataset u ruajt në: {diabetes_csv_path.name}")
print(f"  Rreshta: {len(df_raw)} | Kolona: {len(df_raw.columns)}")
```

## 🔍 Hapi 3: Inspektimi i Dataset-it & EDA (Step 1)
Le të inspektojmë dimensionet, llojet e të dhënave, vlerat unike, statistikën deskriptive dhe balancën e klasave.

```python
print(f"Statistikat bazë për diabetin:")
print(f"Numri i rreshtave: {len(df_raw)}")
print(f"Numri i kolonave : {len(df_raw.columns)}")

print(f"\nTiparet e kolonave dhe vlerat null:")
for col in df_raw.columns:
    print(f"  {col:<26} | tip: {str(df_raw[col].dtype):<8} | null: {df_raw[col].isnull().sum():<3} | zero: {(df_raw[col] == 0).sum()}")

print("\n5 rreshtat e parë:")
display(df_raw.head())

print("\nStatistika deskriptive (me vlerat 0 origjinale):")
display(df_raw.describe().round(2))

print("\nBalanca e klasave ('Outcome'):")
counts = df_raw['Outcome'].value_counts()
for cls, count in counts.items():
    pct = count / len(df_raw) * 100
    print(f"  Klasa {cls} (Jo-Diabetik / Diabetik): {count:>4} pacientë ({pct:.1f}%)")
```

## 🛠️ Hapi 4: Preprocessing i të Dhënave (Step 2)
Në këtë hap kryejmë logjikën kryesore të pastrimit:
1. **Imputimi i Vlerave Mungesë:** Zëvendësojmë vlerat zero (0) në kolonat fizike (`Glucose`, `BloodPressure`, etc.) me medianën e klasës përkatëse.
2. **Capping i Outliers:** Përdorim metodën IQR (Interquartile Range) për t'i kufizuar vlerat ekstreme jashtë kufijve standard pa fshirë rreshta.
3. **Ndarja Train/Test:** Ndarje 80/20 në mënyrë të stratifikuar.
4. **Skalimi i Tipareve:** Përdorim `StandardScaler` për të shmangur *data leakage*.

```python
# 1. Zëvendësimi i vlerave 0
df_clean = df_raw.copy()
cols_with_zero_as_missing = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

for col in cols_with_zero_as_missing:
    df_clean[col] = df_clean[col].astype(float)
    n_zeros = (df_clean[col] == 0).sum()
    if n_zeros > 0:
        for outcome in [0, 1]:
            mask = (df_clean[col] == 0) & (df_clean['Outcome'] == outcome)
            median_val = df_clean.loc[(df_clean[col] != 0) & (df_clean['Outcome'] == outcome), col].median()
            df_clean.loc[mask, col] = median_val
        print(f"  {col:<20}: {n_zeros} vlera 0 -> zëvendësuar me medianën e klasës")

# 2. Trajtimi i outliers me IQR
def handle_outliers_iqr(df: pd.DataFrame, target_col: str, threshold: float = 1.5) -> pd.DataFrame:
    df = df.copy()
    feature_cols = [c for c in df.columns if c != target_col]
    for col in feature_cols:
        if df[col].dtype not in ['int64', 'float64', 'float32'] or df[col].nunique() <= 5:
            continue
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - threshold * IQR
        upper = Q3 + threshold * IQR
        df[col] = df[col].clip(lower, upper)
    return df

df_preprocessed = handle_outliers_iqr(df_clean, 'Outcome')

# 3. Ndarja Train/Test dhe Skalimi
X = df_preprocessed.drop(columns=['Outcome'])
y = df_preprocessed['Outcome']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=RANDOM_STATE
)

scaler = StandardScaler()
X_train_scaled = pd.DataFrame(scaler.fit_transform(X_train), columns=X_train.columns, index=X_train.index)
X_test_scaled = pd.DataFrame(scaler.transform(X_test), columns=X_test.columns, index=X_test.index)

# Ruaj skalerin
scalers = {'diabetes': scaler}
with open(PROCESSED_DIR / "scalers.pkl", 'wb') as f:
    pickle.dump(scalers, f)

# Ruaj të dhënat e procesuara si CSV
train_df = X_train_scaled.copy()
train_df['Outcome'] = y_train.values
test_df = X_test_scaled.copy()
test_df['Outcome'] = y_test.values

train_df.to_csv(PROCESSED_DIR / "diabetes_train.csv", index=False)
test_df.to_csv(PROCESSED_DIR / "diabetes_test.csv", index=False)

print(f"\nPreprocessing përfundoi!")
print(f"  Train Set: {X_train_scaled.shape} | Test Set: {X_test_scaled.shape}")
```

## 🧠 Hapi 5: Trajnimi i Modeleve (Step 3)
Ne do të trajnojmë pesë arkitektura të ndryshme klasifikimi:
1. **k-NN**
2. **Random Forest** (Kryesori)
3. **Logistic Regression**
4. **MLP Arkitektura 1** (Neural Network me 1 shtresë prej 64 neuronesh)
5. **MLP Arkitektura 2** (Neural Network e thellë 128x64x32)

Do të llogarisim metrikat kryesore: **Accuracy, Precision, Recall, F1-Score**.

```python
models = {
    "kNN": KNeighborsClassifier(n_neighbors=5, metric='euclidean', weights='uniform'),
    "Random Forest": RandomForestClassifier(n_estimators=100, max_depth=None, min_samples_split=2, random_state=RANDOM_STATE),
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=RANDOM_STATE, solver='lbfgs'),
    "MLP Arkitektura 1": MLPClassifier(hidden_layer_sizes=(64,), activation='relu', max_iter=500, random_state=RANDOM_STATE, early_stopping=True, validation_fraction=0.1),
    "MLP Arkitektura 2": MLPClassifier(hidden_layer_sizes=(128, 64, 32), activation='relu', max_iter=500, random_state=RANDOM_STATE, early_stopping=True, validation_fraction=0.1, learning_rate='adaptive')
}

results = []

for name, model in models.items():
    # Fit
    model.fit(X_train_scaled, y_train)
    # Predict
    y_pred = model.predict(X_test_scaled)
    
    # Llogarit metrikat
    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, zero_division=0)
    rec = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)
    cm = confusion_matrix(y_test, y_pred)
    
    results.append({
        "dataset": "diabetes",
        "model": name,
        "accuracy": round(acc, 4),
        "precision": round(prec, 4),
        "recall": round(rec, 4),
        "f1_score": round(f1, 4),
        "tn": cm[0,0], "fp": cm[0,1],
        "fn": cm[1,0], "tp": cm[1,1]
    })
    
    # Ruaj modelin e trajnuar
    safe_name = name.lower().replace(" ", "_").replace("(", "").replace(")", "")
    with open(MODELS_DIR / f"diabetes_{safe_name}.pkl", 'wb') as f:
        pickle.dump(model, f)

# Kthe në DataFrame dhe shfaq tabelën krahasuese
df_results = pd.DataFrame(results).sort_values('f1_score', ascending=False)
df_results.to_csv(MODELS_DIR / "results.csv", index=False)

print("TABELA KRAHASUESE E MODELEVE ML (renditur sipas F1-Score):")
display(df_results[['model', 'accuracy', 'precision', 'recall', 'f1_score']])
```

## 🔄 Hapi 6: K-Means Clustering (Step 4)
Clustering na lejon të grupojmë pacientët sipas rrezikut shëndetësor në mënyrë të pavarur (Unsupervised Learning).
Përdorim **Elbow Method** për të konfirmuar numrin e grupeve optimal (K=3) që përputhen me nivelet klinike: *Rrezik i Ulët, Rrezik Mesatar, Rrezik i Lartë*.

```python
X_full = pd.concat([X_train_scaled, X_test_scaled], ignore_index=True)
y_full = pd.concat([y_train, y_test], ignore_index=True)

print("Po ekzekutohet Elbow Method (K=2 deri 10):")
wcss = []
sil = []

for k in range(2, 11):
    km = KMeans(n_clusters=k, random_state=RANDOM_STATE, n_init=10)
    labels = km.fit_predict(X_full)
    wcss.append(km.inertia_)
    sil.append(silhouette_score(X_full, labels))
    print(f"  K={k:>2} | WCSS={km.inertia_:>10.2f} | Silhouette Score={silhouette_score(X_full, labels):.4f}")

# Trajnojmë K-Means me K=3 optimal
km_final = KMeans(n_clusters=3, random_state=RANDOM_STATE, n_init=20)
km_final.fit(X_full)

df_analysis = X_full.copy()
df_analysis['cluster'] = km_final.labels_
df_analysis['outcome'] = y_full.values

# Rendit sipas glukozës mesatare për të lidhur me rrezikun klinik
glucose_means = df_analysis.groupby('cluster')['Glucose'].mean().sort_values()
cluster_order = glucose_means.index.tolist()

risk_labels = {
    cluster_order[0]: 'Rrezik i Ulet',
    cluster_order[1]: 'Rrezik Mesatar',
    cluster_order[2]: 'Rrezik i Larte'
}

cluster_results = []
print(f"\nANALIZA E GRUPEVE TE ZBULUARA:")
for cluster_id in cluster_order:
    subset = df_analysis[df_analysis['cluster'] == cluster_id]
    risk = risk_labels[cluster_id]
    pct_diabetic = subset['outcome'].mean() * 100
    print(f"\n  Cluster {cluster_id} — {risk}:")
    print(f"    Numri i Pacientëve: {len(subset)} ({len(subset)/len(df_analysis)*100:.1f}%)")
    print(f"    Përqindja Diabetik : {pct_diabetic:.1f}%")
    
    cluster_results.append({
        'cluster': cluster_id,
        'risk_level': risk,
        'n_patients': len(subset),
        'pct_diabetic': round(pct_diabetic, 2)
    })

# Ruaj modelin K-Means
with open(MODELS_DIR / "diabetes_kmeans.pkl", "wb") as f:
    pickle.dump({
        'model': km_final,
        'risk_labels': risk_labels,
        'feature_names': X_train.columns.tolist()
    }, f)

pd.DataFrame(cluster_results).to_csv(MODELS_DIR / "kmeans_results.csv", index=False)
```

## ⚙️ Hapi 7: Hyperparameter Tuning (Step 5)
Tuning bën kërkimin e plotë në rrjetë (`GridSearchCV`) me 5-Fold Cross-Validation për të gjetur kombinimin optimal të parametrave për 4 modelet kryesore.

```python
param_grids = {
    "kNN": {
        "model": KNeighborsClassifier(),
        "params": {
            "n_neighbors": [3, 5, 7, 9, 11],
            "weights": ["uniform", "distance"],
            "metric": ["euclidean", "manhattan"]
        }
    },
    "Random Forest": {
        "model": RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=-1),
        "params": {
            "n_estimators": [50, 100, 150],
            "max_depth": [None, 5, 10],
            "min_samples_split": [2, 5, 10]
        }
    },
    "Logistic Regression": {
        "model": LogisticRegression(random_state=RANDOM_STATE, max_iter=1000),
        "params": {
            "C": [0.01, 0.1, 1.0, 10.0],
            "penalty": ["l2"],
            "solver": ["lbfgs", "liblinear"]
        }
    },
    "MLP": {
        "model": MLPClassifier(random_state=RANDOM_STATE, max_iter=500, early_stopping=True),
        "params": {
            "hidden_layer_sizes": [(64,), (128,), (64, 32)],
            "activation": ["relu", "tanh"]
        }
    }
}

tuning_comparison = []

for name, config in param_grids.items():
    print(f"Po kryhet Hyperparameter Tuning për: {name}...")
    grid = GridSearchCV(
        estimator=config["model"],
        param_grid=config["params"],
        cv=5,
        scoring='f1',
        n_jobs=-1
    )
    grid.fit(X_train_scaled, y_train)
    
    best_model = grid.best_estimator_
    y_pred = best_model.predict(X_test_scaled)
    test_f1 = f1_score(y_test, y_pred, zero_division=0)
    
    # Lexo F1 origjinale
    if name == "MLP":
        orig_f1 = df_results[df_results["model"].str.contains("MLP")]["f1_score"].max()
    else:
        orig_f1 = df_results[df_results["model"] == name]["f1_score"].values[0]
        
    improvement = test_f1 - orig_f1
    
    # Ruaj modelin e optimizuar
    safe_name = name.lower().replace(" ", "_")
    with open(MODELS_DIR / f"tuned_diabetes_{safe_name}.pkl", "wb") as f:
        pickle.dump(best_model, f)
        
    tuning_comparison.append({
        "model": name,
        "original_f1": round(orig_f1, 4),
        "tuned_f1": round(test_f1, 4),
        "improvement": round(improvement, 4),
        "best_params": str(grid.best_params_)
    })

df_tuning = pd.DataFrame(tuning_comparison)
df_tuning.to_csv(MODELS_DIR / "tuning_results.csv", index=False)

print("\nKRAHASIMI PAS HYPERPARAMETER TUNING:")
display(df_tuning)
```

## 🛡️ Hapi 8: Stratified K-Fold Cross-Validation (Step 6)
Për të siguruar që modeli i trajnuar ka performancë të qëndrueshme dhe nuk vuan nga *overfitting*, bëjmë Stratified 5-Fold Cross-Validation.

```python
print("Po kryhet Stratified 5-Fold Cross-Validation në Train Set:")
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

cv_results = []
for name, model in models.items():
    scores = cross_val_score(model, X_train_scaled, y_train, cv=skf, scoring='f1', n_jobs=-1)
    print(f"  {name:<22} | F1 Mean: {scores.mean():.4f} ± {scores.std():.4f}  (min={scores.min():.4f}, max={scores.max():.4f})")
    
    cv_results.append({
        "model": name,
        "f1_mean": round(scores.mean(), 4),
        "f1_std": round(scores.std(), 4)
    })

pd.DataFrame(cv_results).to_csv(MODELS_DIR / "cross_validation_results.csv", index=False)
```

## 📊 Hapi 9: Vizualizimet Profesionale (Step 7)
Le të gjenerojmë 5 grafikë interaktivë profesionalë drejtpërdrejt në këtë notebook, të cilët do të ruhen automatikisht në dosjen `visualizations/`.

```python
fig_cm, ax_cm = plt.subplots(figsize=(7, 5))
cm = confusion_matrix(y_test, models["Random Forest"].predict(X_test_scaled))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=['Jo Diabetik', 'Diabetik'], 
            yticklabels=['Jo Diabetik', 'Diabetik'],
            annot_kws={'size': 16, 'weight': 'bold'}, ax=ax_cm)
ax_cm.set_xlabel('Parashikim', fontsize=12, fontweight='bold')
ax_cm.set_ylabel('Reale', fontsize=12, fontweight='bold')
ax_cm.set_title('Confusion Matrix - Random Forest', fontsize=14, fontweight='bold', pad=12)
plt.tight_layout()
plt.savefig(VIS_DIR / "confusion_matrix_diabetes.png", dpi=150)
plt.show()

# 2. Model Comparison
fig_comp, ax_comp = plt.subplots(figsize=(10, 5))
df_results_sorted = df_results.sort_values('f1_score', ascending=False)
metrics = ['accuracy', 'precision', 'recall', 'f1_score']
metric_labels = ['Accuracy', 'Precision', 'Recall', 'F1-Score']
x = np.arange(len(df_results_sorted))
width = 0.2

for i, metric in enumerate(metrics):
    ax_comp.bar(x + (i - 1.5) * width, df_results_sorted[metric], width, label=metric_labels[i], alpha=0.85)
ax_comp.set_xticks(x)
ax_comp.set_xticklabels(df_results_sorted['model'], rotation=15, ha='right')
ax_comp.set_ylabel('Score', fontweight='bold')
ax_comp.set_title('Krahasimi i Performancës së Modeleve ML', fontsize=14, fontweight='bold')
ax_comp.legend(loc='lower right')
ax_comp.set_ylim(0, 1.15)
plt.tight_layout()
plt.savefig(VIS_DIR / "model_comparison.png", dpi=150)
plt.show()

# 3. PCA 2D Cluster
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_full)
fig_pca, axes_pca = plt.subplots(1, 2, figsize=(15, 5))

# Left: Clusters
ax = axes_pca[0]
cluster_colors = ['#2ecc71', '#f39c12', '#e74c3c']
for i, risk_name in enumerate(['Rrezik i Ulet', 'Rrezik Mesatar', 'Rrezik i Larte']):
    c_id = [k for k, v in risk_labels.items() if v == risk_name][0]
    mask = km_final.labels_ == c_id
    ax.scatter(X_pca[mask, 0], X_pca[mask, 1], c=cluster_colors[i], label=risk_name, s=35, alpha=0.6, edgecolors='white')

# Centroids
centroids_pca = pca.transform(km_final.cluster_centers_)
ax.scatter(centroids_pca[:, 0], centroids_pca[:, 1], c='black', marker='X', s=200, edgecolors='white', label='Qendrat')
ax.set_title('K-Means Clusters (3 Nivele Rreziku)', fontsize=13, fontweight='bold')
ax.legend()

# Right: Real Labels
ax = axes_pca[1]
ax.scatter(X_pca[y_full == 0, 0], X_pca[y_full == 0, 1], c='#3498db', label='Jo Diabetik', s=35, alpha=0.6, edgecolors='white')
ax.scatter(X_pca[y_full == 1, 0], X_pca[y_full == 1, 1], c='#e74c3c', label='Diabetik', s=35, alpha=0.6, edgecolors='white')
ax.set_title('Etiketat Reale (Për Krahasim)', fontsize=13, fontweight='bold')
ax.legend()
plt.suptitle('PCA 2D - K-Means Clustering', fontsize=15, fontweight='bold')
plt.tight_layout()
plt.savefig(VIS_DIR / "kmeans_clusters.png", dpi=150)
plt.show()

# 4. Feature Importance (Random Forest)
fig_feat, ax_feat = plt.subplots(figsize=(8, 5))
rf_model = models["Random Forest"]
importances = pd.Series(rf_model.feature_importances_, index=X_train.columns).sort_values()
bars = ax_feat.barh(importances.index, importances.values, color='#3498db', alpha=0.85)
ax_feat.set_xlabel('Rëndësia', fontweight='bold')
ax_feat.set_title('Feature Importance (Random Forest)', fontsize=13, fontweight='bold')
for bar in bars:
    val = bar.get_width()
    ax_feat.text(val + 0.005, bar.get_y() + bar.get_height()/2, f'{val:.3f}', va='center')
plt.tight_layout()
plt.savefig(VIS_DIR / "feature_importance.png", dpi=150)
plt.show()

# 5. Radar Chart
fig_radar, ax_radar = plt.subplots(figsize=(7, 7), subplot_kw=dict(projection='polar'))
angles = np.linspace(0, 2 * np.pi, len(metrics), endpoint=False).tolist()
angles += angles[:1]
colors = plt.cm.viridis(np.linspace(0, 0.9, 5))

for i, (_, row) in enumerate(df_results.iterrows()):
    values = [row[m] for m in metrics]
    values += values[:1]
    ax_radar.plot(angles, values, 'o-', linewidth=2, label=row['model'], color=colors[i])
    ax_radar.fill(angles, values, alpha=0.1, color=colors[i])
    
ax_radar.set_xticks(angles[:-1])
ax_radar.set_xticklabels(metric_labels, fontsize=11, fontweight='bold')
ax_radar.set_ylim(0, 1.1)
ax_radar.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))
ax_radar.set_title('Krahasimi Radial i Modeleve', fontsize=13, fontweight='bold', pad=15)
plt.tight_layout()
plt.savefig(VIS_DIR / "metrics_radar.png", dpi=150)
plt.show()
```

## 🤖 Hapi 10: Ndërfaqja e Parashikimit në Production (Step 8)
Le të krijojmë një funksion parashikimi të integruar që merr të dhënat e një pacienti, bën skalimin, kryen klasifikimin me modelin Random Forest dhe K-Means, dhe kthen një format JSON të pastër për backend-in. Do ta testojmë këtë me dy pacientë shembull.

```python
def predict_diabetes_risk(patient_data: dict) -> dict:
    # Rendi i fushave klinike
    features = ["Pregnancies", "Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI", "DiabetesPedigreeFunction", "Age"]
    
    # Konverto në DataFrame
    df_patient = pd.DataFrame([[patient_data[f] for f in features]], columns=features)
    
    # Skalo tiparet
    df_patient_scaled = pd.DataFrame(scaler.transform(df_patient), columns=features)
    
    # Parashiko klasën dhe probabilitetin
    prediction = int(models["Random Forest"].predict(df_patient_scaled)[0])
    probability = float(models["Random Forest"].predict_proba(df_patient_scaled)[0][1])
    
    # Përcakto nivelin e rrezikut
    if probability < 0.3:
        risk_level = "I ulet"
    elif probability < 0.6:
        risk_level = "Mesatar"
    else:
        risk_level = "I larte"
        
    # Përcakto grupin K-Means
    cluster_id = int(km_final.predict(df_patient_scaled)[0])
    risk_group = risk_labels.get(cluster_id, "E panjohur")
    
    return {
        "prediction": prediction,
        "probability": round(probability, 4),
        "risk_level": risk_level,
        "risk_group": risk_group,
        "message": f"Probabiliteti i diabetit: {probability*100:.1f}%. ({risk_level}). Grupi K-Means: {risk_group}"
    }

# Test 1: Pacient me rrezik të lartë
pacient_rrezik_larte = {
    "Pregnancies": 6, "Glucose": 152, "BloodPressure": 82, "SkinThickness": 39, "Insulin": 220, "BMI": 36.8, "DiabetesPedigreeFunction": 0.725, "Age": 45
}
print("TEST 1: Pacient me Rrezik të Lartë")
print(predict_diabetes_risk(pacient_rrezik_larte))

# Test 2: Pacient i shëndetshëm
pacient_rrezik_ulet = {
    "Pregnancies": 1, "Glucose": 88, "BloodPressure": 62, "SkinThickness": 21, "Insulin": 0, "BMI": 23.4, "DiabetesPedigreeFunction": 0.284, "Age": 23
}
print("\nTEST 2: Pacient i Shëndetshëm")
print(predict_diabetes_risk(pacient_rrezik_ulet))
```
